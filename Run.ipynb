# Data loading and preprocessing


```python
import pandas as pd

data_scale = 100000

def preprocess(file:str):
	filename = "datasets/" + file
	df = pd.read_csv(filename, sep='\t', header=None)
	df.columns = ['userid','venid','vencatid','venname','lat','long','tz','time']
	df = df.drop(['vencatid', 'venname', 'lat', 'long', 'tz', 'time'], axis=1)
	index = df.index
	number_of_rows = len(index)
	df = df.drop(range(data_scale, number_of_rows))
	df['val'] = 1
	df.drop_duplicates(inplace=True)
	nr_users = df['userid'].nunique()
	df = df.groupby(['venid'], as_index=False)['val'].sum()

	file_w = open(file, 'w')
	file_w.write(str(nr_users) + '\n')
	i = 0
	for item in df.iterrows():
		file_w.write(df.iloc[i][0]+'\t'+str(df.iloc[i][1])+'\n')
		i = i + 1

preprocess('server.txt')
preprocess('client.txt')
```
